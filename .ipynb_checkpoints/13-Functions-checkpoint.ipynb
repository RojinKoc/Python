```python
def summation(num1,num2,num3):
    print(num1+num2+num3)
```

```python
summation(10,20,30)
```

```python
#args, kwargs (arguments, key word arguments)
```

```python
def args_sum(*args):
    return sum(args)
```

```python
args_sum(10,20,30,40)
```

```python
def args_example(*args):
    print(args)
```

```python
args_example(23,23432,546,7786,345)
```

```python
def kwargs_example(**kwargs):
    print(kwargs)
```

```python
kwargs_example(apple=100,banana=150,melon=200)
```

```python
def kwargs_example_2(**kwargs):
    if "apple" in kwargs:
        print("applee")
    else:
        print(":(:")
```

```python
kwargs_example_2(apple=100,banana=150,melon=200)
```

```python
kwargs_example_2(banana=150,melon=200)
```
