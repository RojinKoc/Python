```python
class Person():
    #property
    name = ""
    age = 0
    gender = ""

    #method, initializer
    def __init__(self,name,age,gender):
        self.name = name
        self.age = age
        self.gender = gender
```

```python
class Person():
    #method, initializer
    def __init__(self,name,age,gender):
        self.name = name
        self.age = age
        self.gender = gender
    def printName(self):
        print(self.name)
        
```

```python
rojin = Person("Rojin",23,"Female")
```

```python
rojin.printName()
```
