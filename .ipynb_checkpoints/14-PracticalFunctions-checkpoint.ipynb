```python
def divideNumber(number):
    return number / 2
```

```python
divideNumber(20)
```

```python
myList = [3,5,7,10,20,30]
```

```python
myResultList = []
for num in myList:
    myResultList.append(divideNumber(num))
print(myResultList)
```

```python
#map
```

```python
help(map)
```

```python
list(map(divideNumber,myList))
```

```python
def controlString(string):
    return "Rojin" in string
```

```python
controlString("Rojin Koç")
```

```python
myStringList = ["Rojin", "Rojin Koç", "Koç"]
```

```python
list(map(controlString,myStringList))
```

```python
#filter
```

```python
list(filter(controlString,myStringList))
```

```python
#LEGB -> local enclosing global built-in
```

```python
#Global
myString = "Rojin"
print(myString)
def myFunction():
    #Enclosing
    myString = "Rojin 2"
    print(myString)

    def myFunction2():
        #Local
        myString = "Rojin 3"
        print(myString)

    myFunction2()
print(myString)
```

```python
myFunction()
```

```python
y = 10
def newFunction(y):
    print(y)
    y = 5
    print(y)
    return y
```

```python
y = 10
def changeY():
    global y
    y = 5
    print(y)
```
